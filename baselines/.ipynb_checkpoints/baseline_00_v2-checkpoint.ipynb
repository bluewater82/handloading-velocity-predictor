- Dataset has grown 20x since previous baselines were run. I must start over with the basic models to make sure nothing within the dataset was broken between then and now.
- As with previous ZERO baseline, this is simply meant to find the absolute mae floor based on pure model ignorance. *Only velocities will be looked at.*
- Updated with standard OLS model to try standardizing results for clean comparisons between baselines

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("308_baseline_data.csv")
df.head()
```

```python
target = df["velocity_fps"].astype(float)
mean_velocity = target.mean()
print(f"Total mean velocity across dataset: {mean_velocity:.2f} ft/s.")
```

```python
baseline_zero_predictions = np.full_like(target, mean_velocity)
baseline_zero_mae = np.mean(np.abs(target - baseline_zero_predictions))
print(f"Floor baseline w/ no features: {baseline_zero_mae:.2f} ft/s.")
```
