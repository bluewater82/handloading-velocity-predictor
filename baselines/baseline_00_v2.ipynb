### baseline ZERO.v2

- Dataset has grown 20x since previous baselines were run. I must start over with the basic models to make sure nothing within the dataset was broken between then and now.
- As with previous ZERO baseline, this is simply meant to find the absolute mae floor based on pure model ignorance. *Only velocities will be looked at.*

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("baseline_data_updated.csv")
df.head(20)
```

```python
target = df["velocity_fps"].astype(float)
mean_velocity = target.mean()
mean_velocity
```

```python
baseline_zerov2_predictions = np.full_like(target, mean_velocity)
baseline_zerov2_mae = np.mean(np.abs(target - baseline_zerov2_predictions))
baseline_zerov2_mae
```

### Absolute ignorance mae based on new data is 262.739. This is our new floor moving forward. All future models should perform better than this
