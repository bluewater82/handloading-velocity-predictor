### baseline ZERO: "Dumb" reference
- predict mean velocity for every row

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("baseline_zero_data.csv")
df.head(20)
```

```python
target = df["velocity_fps"].astype(float)  # define target and cast as float
```

```python
mean_velocity = target.mean()  # this is our dumb predictor, mean value only across all rows
```

```python
mean_velocity
```

```python
baseline_zero_predictions = np.full_like(target, mean_velocity)
baseline_zero_mae = np.mean(np.abs(target - baseline_zero_predictions))
baseline_zero_mae
```

### baseline zero mae = 774.224 = absolute floor
- This is the worst the model should behave as it has zero features to work with. If future models yield worse errors then something in them is broken.
